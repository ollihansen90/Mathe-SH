## 🧐 Unsere Aufgabe

1. Ersetze einzelne Wörter oder Buchstabenfolgen mit `.replace()`.
2. Sammle die Ersetzungen in einem Dictionary.
3. Sorge dafür, dass längere Buchstabenfolgen immer zuerst ersetzt werden.
4. Erweitere das Wörterbuch mit eigenen Ideen.
5. Teste dein Programm an möglichst langen Wörtern!

**Bonus:** 🏆 Kann man ein Wort so verschlüsseln, dass es kaum noch zu lesen ist?

```python
emojidict = {
             "schiff": "🚢",
             "dampf": "☁️",
             "kapitän": "👨‍✈️",
             "mütze": "🧢",
             "schaf": "🐑",
}

wort = "Donaudampfschifffahrtsgesellschaftskapitänsmütze"
def emojify(wort, emojis):
    wort = wort.lower()
    liste = sorted(emojis, key=len, reverse=True)

    for teilwort in liste:
        wort = wort.replace(teilwort,emojis[teilwort])

    return wort

print(emojify(wort, emojidict))
```
