## 🔄 Padding – Was passiert am Rand?

Auch unsere 2D-Welt hat einen Rand. Eine Zelle ganz links hat zum Beispiel keinen Nachbarn mehr links von sich.

Dafür können wir mit `np.pad()` einen zusätzlichen Rand um unsere Welt legen:

```python
np.pad(welt, 2, mode="wrap")
```

Dabei bedeutet:

- `welt` → unser Array
- `2` → Breite des zusätzlichen Randes
- `mode` → bestimmt, **wie der neue Rand gefüllt wird**

### 🧱 Verschiedene Möglichkeiten für `mode`

NumPy bietet verschiedene Arten von Padding:

- `mode="constant"` → der Rand wird mit einem festen Wert gefüllt, standardmäßig mit `0`
- `mode="mean"` → der Rand wird aus dem **Mittelwert** der Werte berechnet
- `mode="edge"` → die Werte am Rand werden **wiederholt**
- `mode="wrap"` → die gegenüberliegende Seite der Welt wird **fortgesetzt** (periodisch)

### 🌍 Was verwenden wir?

Für unser **Game of Life** verwenden wir:

```python
mode="wrap"
```

Damit erhalten wir wieder eine **periodische Randbedingung**:

👉 Was auf einer Seite aus unserer Welt herausgeht, erscheint auf der gegenüberliegenden Seite wieder.

### 🔍 Probiert es aus

👉 **ob die Zelle selbst vorher lebendig oder tot war.**

Eine **tote Zelle** trägt selbst `0` zur Summe bei. Ihre Summe liegt deshalb zwischen `0` und `8`.

Eine **lebende Zelle** trägt dagegen zusätzlich `8` bei. Ihre Summe liegt deshalb zwischen `8` und `16`.

Zum Beispiel:

- `3` → tote Zelle mit **3 lebenden Nachbarn**

Damit lassen sich die Regeln von **Conways Game of Life** sehr einfach formulieren:

- Eine tote Zelle mit genau **3 Nachbarn** wird lebendig.
- Eine lebende Zelle mit **2 oder 3 Nachbarn** bleibt lebendig.
