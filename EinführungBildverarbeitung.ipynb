## Digitale Bilder 🦆

Bisher haben wir mit NumPy hauptsächlich Zahlen und Tabellen verarbeitet. Heute wollen wir uns anschauen, wie ein Computer **digitale Bilder** darstellt und wie wir diese mit NumPy verändern können!

Ein digitales Bild besteht aus vielen kleinen Bildpunkten, den sogenannten **Pixeln**. Jeder Pixel besitzt eine Farbe, die durch drei Zahlen beschrieben werden kann:

- 🔴 **Rot (R):** Wie viel Rot enthält der Pixel?
- 🟢 **Grün (G):** Wie viel Grün enthält der Pixel?
- 🔵 **Blau (B):** Wie viel Blau enthält der Pixel?

Durch die Kombination dieser drei Farbkanäle lassen sich Millionen unterschiedlicher Farben darstellen.

Normalerweise liegen die Farbwerte zwischen **0 und 255**. Für unsere Berechnungen verwenden wir stattdessen Zahlen zwischen **0 und 1**.

Ein digitales Farbbild lässt sich deshalb wunderbar als dreidimensionales NumPy-Array darstellen! Die drei Dimensionen entsprechen der **Höhe**, der **Breite** und den **Farbkanälen** des Bildes.

Schauen wir uns ein Beispiel an! 🦆

Mit `io.imread()` laden wir ein Bild aus dem Internet. Anschließend wandeln wir die Farbwerte in Fließkommazahlen um und verkleinern das Bild ein wenig, indem wir nur jeden zweiten Pixel in jeder Richtung auswählen.

Mit `plt.imshow()` können wir uns das Bild anzeigen lassen. Zum Schluss verrät uns `bild.shape`, wie groß unser Array ist.

```python
import numpy as np
from skimage import io
import matplotlib.pyplot as plt

every = 2
bild = io.imread("https://www.planet-wissen.de/natur/voegel/entenvoegel/stockentensindguteflieger100~_v-gseapremiumxl.jpg")
bild = bild.astype(float)
bild = bild[::every,::every]
bild = bild/np.max(bild)

plt.figure()
plt.imshow(bild)
plt.show()

print(bild.shape)
```

## Die drei Farbkanäle: Rot, Grün und Blau 🌈

Ein Farbbild besteht aus drei Kanälen:

- 🔴 **Rotkanal**
- 🟢 **Grünkanal**
- 🔵 **Blaukanal**

Für jeden Pixel speichern diese drei Kanäle, wie stark die jeweilige Farbe vertreten ist. Wenn ein Pixel zum Beispiel einen hohen Rotwert, aber kleine Grün- und Blauwerte besitzt, wirkt er eher rötlich.

Um die einzelnen Kanäle sichtbar zu machen, können wir jeweils die beiden anderen Farbkanäle auf `0` setzen. So sehen wir, **welcher Teil des Bildes besonders viel Rot, Grün oder Blau enthält**.

Schauen wir uns unser Entenbild einmal kanalweise an!

```python
plt.figure(figsize=[15,5])

plt.subplot(1,3,1)
rot = bild.copy()
rot[:,:,[1,2]] = 0
plt.imshow(rot)
plt.title("Rotkanal")

plt.subplot(1,3,2)
gruen = bild.copy()
gruen[:,:,[0,2]] = 0
plt.imshow(gruen)
plt.title("Grünkanal")

plt.subplot(1,3,3)
blau = bild.copy()
blau[:,:,[0,1]] = 0
plt.imshow(blau)
plt.title("Blaukanal")

plt.show()

for i in range(3):
    print(["Rot:", "Grün:", "Blau:"][i], np.mean([rot, gruen, blau][i]))
```

## Aus drei Farben wird eine! ⚫⚪

Jetzt wollen wir unser buntes Entenbild in ein **Grauwertbild** umwandeln!

Dafür berechnen wir für jeden Pixel einen neuen Helligkeitswert, indem wir die drei Farbkanäle miteinander mischen.

Wie stark die einzelnen Farben berücksichtigt werden, bestimmen wir mit drei **Gewichten**. Dabei soll die Summe aller Gewichte genau `1` ergeben. Eine solche Mischung nennt man **Konvexkombination**.

Wenn wir alle drei Farbkanäle gleich stark gewichten, erhalten wir einfach ihren Durchschnitt!

**Aufgabe:** 🦆

Verändere die Werte von `rotanteil` und `gruenanteil` und beobachte, wie sich das Grauwertbild verändert.

- Was passiert, wenn du nur den Rotkanal verwendest?
- Was passiert, wenn du den Grünkanal besonders stark gewichtest?
- Welche Kombination liefert deiner Meinung nach das schönste Grauwertbild?

**Achtung:** Alle drei Gewichte müssen zwischen `0` und `1` liegen! Der Blauanteil wird automatisch berechnet. `rotanteil` und `gruenanteil` dürfen zusammen nicht größer als `1` sein

Damit wir unser Ergebnis wieder als RGB-Bild darstellen können, kopieren wir den berechneten Grauwert mit `np.stack()` in alle drei Farbkanäle. Wenn Rot, Grün und Blau denselben Wert besitzen, entsteht ein Grauton!

```python
rotanteil = 0
gruenanteil = 0
assert rotanteil+gruenanteil<=1, "Rotanteil und Grünanteil sind zu groß."
blauanteil = 1-rotanteil-gruenanteil

grauwertbild = rotanteil*bild[:,:,0]+gruenanteil*bild[:,:,1]+blauanteil*bild[:,:,2]
grauwertbild = np.stack(3*[grauwertbild], axis=-1)

plt.figure()
plt.imshow(grauwertbild, vmin=0, vmax=1)
plt.show()
```

## Sind alle Farben gleich hell? 👀

Vielleicht ist dir aufgefallen, dass manche Kombinationen der Farbkanäle natürlicher aussehen als andere.

Das liegt daran, dass unsere Augen **nicht für alle Farben gleich empfindlich** sind! Grünes Licht nehmen wir bei gleicher physikalischer Lichtleistung beispielsweise deutlich heller wahr als blaues Licht.

Deshalb verwendet man bei der Umwandlung von Farbbildern in Grauwertbilder häufig spezielle Gewichte:

- 🔴 **Rot:** 21,26 %
- 🟢 **Grün:** 71,52 %
- 🔵 **Blau:** 7,22 %

Diese Werte stammen aus einem internationalen Standard für die Darstellung von Farben (Rec. 709). Sie berücksichtigen sowohl die Eigenschaften unserer Augen als auch die festgelegten RGB-Grundfarben.

Mit diesen Gewichten können wir unser Entenbild in ein Grauwertbild umwandeln, bei dem die Helligkeitsunterschiede besser berücksichtigt werden. 🦆

```python
rotanteil = 0.2126
gruenanteil = 0.7152
blauanteil = 1-rotanteil-gruenanteil # = 0.0722

grauwertbild = rotanteil*bild[:,:,0]+gruenanteil*bild[:,:,1]+blauanteil*bild[:,:,2]
grauwertbild = np.stack(3*[grauwertbild], axis=-1)

plt.figure()
plt.imshow(grauwertbild)
plt.show()
```

## Das Zwischenbild: Farbe und Grauwert mischen 🎨

Bisher haben wir zwei Versionen unseres Bildes kennengelernt:

- das **Originalbild** mit allen Farben
- das **Grauwertbild** ohne Farbinformation

Jetzt können wir beide Bilder miteinander mischen!

Dazu bilden wir wieder eine **Konvexkombination**:

`zwischenbild = anteil * bild + (1 - anteil) * grauwertbild`

Der Wert `anteil` bestimmt, wie bunt das Ergebnis noch ist:

- `anteil = 0` → nur das Grauwertbild
- `anteil = 1` → das Originalbild
- `0 < anteil < 1` → ein **entsättigtes** Bild, also ein Bild mit schwächeren Farben

Je kleiner `anteil` ist, desto grauer wirkt das Bild.

**Aufgabe:** Probiere verschiedene Werte für `anteil` aus!

- Wie sieht das Bild bei `anteil = 0.5` aus?
- Ab welchem Wert wirkt das Bild fast schon grau?
- Was passiert bei `anteil = 1`?

```python
anteil = 0.5
zwischenbild = anteil*bild+(1-anteil)*grauwertbild

plt.figure(figsize=[10,5])
plt.subplot(1,2,1)
plt.imshow(bild)
plt.title("Original")
plt.subplot(1,2,2)
plt.imshow(np.clip(zwischenbild, a_min=0, a_max=1))
plt.title("Anteil = "+str(anteil))
plt.show()
```

## Bonus: Noch bunter als das Original? 🌈

Spannend wird es, wenn wir **keine** Konvexkombination mehr verwenden, also zum Beispiel `anteil > 1` wählen.

Dann ist der Faktor vor dem Grauwertbild negativ, und das Bild wird nicht mehr „in Richtung grau“, sondern **von grau weg** verschoben. Dadurch werden die Farben oft **kräftiger** — das Bild wirkt also stärker gesättigt.

Zum Beispiel:

- `anteil = 1.2`
- `anteil = 1.5`

können zu besonders leuchtenden Farben führen.

Allerdings können dabei Pixelwerte kleiner als `0` oder größer als `1` entstehen. Deshalb verwenden wir `np.clip(...)`, damit alle Werte wieder im gültigen Bereich liegen.

**Zusatzaufgabe:** Teste auch Werte außerhalb des Intervalls `[0,1]`.

- Was passiert bei `anteil = 10`?
- Was passiert bei `anteil = -0.5`?
- Warum ist `np.clip()` hier nützlich?

```python
anteil = 1.7
zwischenbild = anteil*bild+(1-anteil)*grauwertbild

plt.figure(figsize=[10,5])
plt.subplot(1,2,1)
plt.imshow(bild)
plt.title("Original")
plt.subplot(1,2,2)
plt.imshow(np.clip(zwischenbild, a_min=0, a_max=1))
plt.title("Anteil = "+str(anteil))
plt.show()
```
