```python
import matplotlib.pyplot as plt

regeln = {
    "000": 0,
    "001": 1,
    "010": 1,
    "011": 1,
    "100": 1,
    "101": 0,
    "110": 0,
    "111": 0,
}
groesse = 111
n_epochs = 100

welt = []
for i in range(groesse):
    welt.append(0)

welt[groesse//2] = 1

verlauf = [welt]
def eltern(w, index):
    #eintraege = "".join([str(welt[index-1]), str(welt[index]), str(welt[index+1])])
    eintraege = str(w[index-1])+str(w[index])+str(w[(index+1)%len(w)])
    return eintraege

for t in range(n_epochs):
    neue_welt = []
    for i in range(groesse):
        neue_welt.append(regeln[eltern(verlauf[-1],i)])
    verlauf.append(neue_welt)

plt.figure()
plt.imshow(verlauf)
plt.show()
```
