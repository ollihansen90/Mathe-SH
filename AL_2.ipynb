```python
import matplotlib.pyplot as plt

regeln = {
    "000": 0,
    "001": 1,
    "010": 1,
    "011": 1,
    "100": 1,
    "101": 0,
    "110": 0,
    "111": 0,
}
```

```python
groesse = 500
n_epochs = 500

welt = []
for i in range(groesse):
    welt.append(0)

welt[5] = 1

verlauf = [welt]
def eltern(w, index):
    #eintraege = "".join([str(welt[index-1]), str(welt[index]), str(welt[index+1])])
    eintraege = str(w[index-1])+str(w[index])+str(w[(index+1)%len(w)])
    return eintraege

```

```python
for t in range(n_epochs):
    neue_welt = []
    for i in range(groesse):
        neue_welt.append(regeln[eltern(verlauf[-1],i)])
    verlauf.append(neue_welt)

plt.figure()
plt.imshow(verlauf)
plt.show()
```

```python
class World:
    def __init__(self,g,times):
        self._ = [[0 for _ in range(times)] for _ in range(g)]
    def setlive(self,posx,posy):
        self._[posy][posx] = 1
    def parents(self,pos: tuple):
        return ''.join([str(self._[pos[1]-1][i+pos[0]]) for i in range(3)])
welt = World(11,10)
welt.setlive(4,0)
print(welt.parents((4,0)))
```
